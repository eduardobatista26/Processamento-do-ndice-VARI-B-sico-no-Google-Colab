```python
from google.colab import files
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Upload do arquivo de imagem do computador para o ambiente do Colab
uploaded = files.upload()
filename = list(uploaded.keys())[0]

def analisar_vigor_estresse(caminho_imagem):
    # 2. Ler a imagem em BGR e converter os valores de pixel para float (0.0 a 1.0)
    img = cv2.imread(caminho_imagem).astype(float) / 255.0

    # Separar os canais de cor RGB (OpenCV lê na ordem B, G, R)
    B, G, R = cv2.split(img)

    # 3. Calcular a fórmula do VARI: (Verde - Vermelho) / (Verde + Vermelho - Azul)
    # O termo 1e-6 (0.000001) evita erro de divisão por zero
    vari = (G - R) / (G + R - B + 1e-6)

    # 4. Normalizar a matriz do índice para a escala de imagem de 8 bits (0 a 255)
    vari_norm = cv2.normalize(vari, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    # 5. Aplicar o mapa de cores 'JET' (Pseudocor para interpretação visual)
    heatmap = cv2.applyColorMap(vari_norm, cv2.COLORMAP_JET)

    # 6. Exibir a imagem original e o mapa de calor lado a lado
    plt.figure(figsize=(16, 8))

    plt.subplot(1, 2, 1)
    plt.title("Imagem Original", fontsize=15)
    plt.imshow(cv2.cvtColor((img * 255).astype(np.uint8), cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.title("Mapa de Vigor/Estresse (VARI)", fontsize=15)
    plt.imshow(cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.show()

    # Exibir a média global do índice na imagem
    print(f"\n--- Resultado da Fenotipagem Digital ---")
    print(f"Índice de Vigor Médio da Amostra: {np.mean(vari):.4f}")

# Executar a função com a imagem carregada
analisar_vigor_estresse(filename)
```
